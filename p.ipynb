```python
class User:
    def __init__(self, name):
        self.name = name

    def apply_leave(self, days):
        print(self.name, "applied for", days, "days leave")


class Admin:
    def approve_leave(self, user):
        print("Leave approved for", user.name)


user = User("Kailash")
admin = Admin()

user.apply_leave(3)
admin.approve_leave(user)
```
